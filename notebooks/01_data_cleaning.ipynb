```python
df["city"] = df["city"].str.strip().str.replace(", krfr. Stadt", "", regex=False)   # Leerzeichen und den Zusatz „, krfr. Stadt“ entfernen
```

```python
df["year"] = df["year"].astype(int)  # Jahr in Ganzzahl umwandeln
df["year"].dtype  
```

```python
df["arrivals"] = df["arrivals"].replace("-", "0") 
df["arrivals_yoy_pct"] = df["arrivals_yoy_pct"].replace("-", "0")
df["overnight_stays"] = df["overnight_stays"].replace("-", "0")
df["overnight_stays_yoy_pct"] = df["overnight_stays_yoy_pct"].replace("-", "0")

# „-“ durch „0“ ersetzen
```

```python
(df == "-").sum() 
```

```python
# Ankünfte und Übernachtungen in ganze Zahlen umwandeln
df["arrivals"] = df["arrivals"].astype(int)
df["overnight_stays"] = df["overnight_stays"].astype(int)
```

```python
df[["arrivals", "overnight_stays"]].dtypes
```

```python
df["average_stay"] = df["average_stay"].replace("-", float("nan"))  # „-“ durch NaN ersetzen 
# Wir setzen nicht 0 ein, denn „der Durchschnitt lässt sich nicht berechnen“ ist nicht dasselbe wie „der Durchschnitt beträgt 0 Nächte“.
# float("nan") → ein tatsächlicher NaN-Wert, der in einer numerischen Spalte verwendet werden kann.
```

```python
# Dezimalkommas durch Punkte ersetzen
df["average_stay"] = df["average_stay"].str.replace(",", ".", regex=False)  
```

```python
df["average_stay"].dtype
```

```python
# In Dezimalzahlen umwandeln
df["average_stay"] = df["average_stay"].astype(float)  
```

```python
df["average_stay"].dtype
```

```python
# Fehlende Werte in der Aufenthaltsdauer zählen
df["average_stay"].isna().sum()  
```

```python
df.loc[df["average_stay"].isna(),"arrivals"].value_counts()  
# Ankünfte für Zeilen mit fehlender Aufenthaltsdauer prüfen
```

```python
month_numbers = {
    "Januar": 1,
    "Februar": 2,
    "März": 3,
    "April": 4,
    "Mai": 5,
    "Juni":6,
    "Juli":7,
    "August": 8,
    "September": 9,
    "Oktober": 10,
    "November": 11,
    "Dezember": 12
}
```

```python
df["month_number"] = df["month"].map(month_numbers) #Monatsnummern für die richtige Sortierung in SQL und Power BI erstellen
```

```python
df[["month", "month_number"]].head(12)
```

```python
df["date"] = pd.to_datetime({
    "year" : df["year"],
    "month" : df["month_number"],
    "day" : 1
})

# Monatsdatum für Zeitreihen in SQL und Power BI erstellen
```

```python
df["arrivals_yoy_pct"] = df["arrivals_yoy_pct"].replace("x", float("nan"))  # „x“ durch NaN ersetzen
```

```python
df["overnight_stays_yoy_pct"] = df["overnight_stays_yoy_pct"].replace("x", float("nan"))  # „x“ durch NaN ersetzen
```

```python
df["arrivals_yoy_pct"] = df["arrivals_yoy_pct"].str.replace(",", ".").astype(float)
df["overnight_stays_yoy_pct"] = df["overnight_stays_yoy_pct"].str.replace(",", ".").astype(float)
```

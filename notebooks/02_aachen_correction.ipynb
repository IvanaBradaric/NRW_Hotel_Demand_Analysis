```python
df_aachen['city'] = df_aachen['city'].str.replace(", krfr. Stadt", "", regex=False)
```

```python
df_aachen['city'] = df_aachen['city'].str.replace('(ab 21.10.2009)', '', regex=False).str.strip()
```

```python
df_aachen['arrivals'] = df_aachen['arrivals'].replace("-", "0")
df_aachen['arrivals_yoy_pct'] = df_aachen['arrivals_yoy_pct'].replace("-", "0")
df_aachen['overnight_stays'] = df_aachen['overnight_stays'].replace("-", "0")
df_aachen['overnight_stays_yoy_pct'] = df_aachen['overnight_stays_yoy_pct'].replace("-", "0")
df_aachen['average_stay'] = df_aachen['average_stay'].replace("-", float("nan"))
df_aachen['arrivals_yoy_pct'] = df_aachen['arrivals_yoy_pct'].replace("x", float("nan"))
df_aachen['overnight_stays_yoy_pct'] = df_aachen['overnight_stays_yoy_pct'].replace("x", float("nan"))

```

```python
df_aachen.isna().sum()
```

```python
df_aachen["year"] = df_aachen["year"].astype(int)
df_aachen["arrivals"] = df_aachen["arrivals"].astype(int)
df_aachen["overnight_stays"] = df_aachen["overnight_stays"].astype(int)
```

```python
df_aachen.dtypes
```

```python
df_aachen["arrivals_yoy_pct"] = df_aachen["arrivals_yoy_pct"].replace(",", ".", regex=True).astype(float)
df_aachen["overnight_stays_yoy_pct"] = df_aachen["overnight_stays_yoy_pct"].replace(",", ".", regex=True).astype(float)
df_aachen["average_stay"] = df_aachen["average_stay"].replace(",", ".", regex=True).astype(float)
```

```python
df_aachen.dtypes
```

```python
month_numbers = {
    "Januar": 1,
    "Februar": 2,
    "März": 3,
    "April": 4,
    "Mai": 5,
    "Juni":6,
    "Juli":7,
    "August": 8,
    "September": 9,
    "Oktober": 10,
    "November": 11,
    "Dezember": 12
}
```

```python
df_aachen["month_number"] = df_aachen["month"].map(month_numbers)
```

```python
df_aachen[["month","month_number"]].head(12)
```

```python
df_aachen["date"] = pd.to_datetime({
    "year" : df_aachen["year"],
    "month" : df_aachen["month_number"],
    "day" : 1
})
```
